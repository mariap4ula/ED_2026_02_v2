## Ejercicio 1 — R-4.1: Producto recursivo

Describa e implemente un algoritmo recursivo para calcular el producto de dos enteros positivos $m$ y $n$, usando **solo suma y resta** (no puede usar el operador `*`).

Programe una función `producto_recursivo(m, n)`.

```python
def producto_recursivo(m, n):
    '''Función recursiva para el producto de enteros'''
    if n == 1:
        return m
    if n < 0:
        return -producto_recursivo(m, -n)
    return m + producto_recursivo(m, n-1)
```

```python
producto_recursivo(9,-7)
```

```python
print(producto_recursivo(6, 7), "esperado:", 6*7)
print(producto_recursivo(0, 9), "esperado:", 0*9)
```

## Ejercicio 3 — R-4.5: Logaritmo base 2 entero

Describa e implemente un algoritmo recursivo para calcular la parte entera del logaritmo en base 2 de $n$ (con $n \geq 1$), usando **solo división entera y suma** (no puede usar `math.log`).

Programe `log2_entero(n)`.

```python
def log2_entero(n):
    if n < 2:
        return 0
    return 1 + log2_entero(n/2)
```

```python
import math
for n in [1,2,4,8,16,32,64, 65, 128, 129]:
    print(n, "->", log2_entero(n), "esperado:", math.log2(n))
```

```python
# Nota: goodrich/ch04/unique_bad.py tiene un bug (llama a `unique` en vez de
# `unique3` dentro de sí misma) y lanza NameError al ejecutarse. Se reproduce
# aquí una copia corregida, con la misma idea de recursión múltiple solapada,
# solo para poder comparar tiempos de ejecución.
def unique3(S, start, stop):
    """Return True if there are no duplicate elements in slice S[start:stop]."""
    if stop - start <= 1: return True
    elif not unique3(S, start, stop-1): return False
    elif not unique3(S, start+1, stop): return False
    else: return S[start] != S[stop-1]

print(unique_eficiente([1, 2, 3, 4]), unique3([1,2,3,4], 0, 4))
print(unique_eficiente([1, 2, 2, 4]), unique3([1,2,2,4], 0, 4))
```

```python
from time import time

S = list(range(20))

a = time()
r1 = unique3(S, 0, len(S))
b = time()
print(f"unique3 (recursión múltiple solapada): {r1}  {b-a:.4f} s")

a = time()
r2 = unique_eficiente(S)
b = time()
print(f"unique_eficiente (orden + recursión):  {r2}  {b-a:.4f} s")
```
